```python
# Create a dataset
# Example we are taking daily electricity consumption of 20 houses
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

data = np.array([4.1,4.3,4.8,5.0,5.2,5.5,6.0,6.2,4.9,5.1,5.3,5.8,6.1,4.7,5.4,5.6,4.6,6.3,5.9,5.0])
```

```python
# Step 1: Calculate the mean and variance of the data
mean = np.mean(data)
variance = np.var(data)

print(f"Mean: {mean}")
print(f"Variance: {variance}")
```

```python
x = np.linspace(min(data)-1, max(data)+1, 100)
pdf = norm.pdf(x, mean, np.sqrt(variance))
plt.hist(data, bins=10, density=True, alpha=0.6, color='g', label='Data Histogram')
plt.plot(x, pdf, 'k', linewidth=2, label='Gaussian PDF')
plt.title('Univariate Gaussian Distribution')
plt.xlabel('Electricity Consumption (kWh)')
plt.ylabel('Density')
plt.legend()
plt.show()
```

```python
# Task 2: Understand covariance and elleptical distribution
# Understand covariance and elleptical distribution
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import multivariate_normal

np.random.seed(42)

# Generate data
daily = np.random.normal(5, 0.5, 100)  # Daily consumption
peak = np.random.normal(2, 0.3, 100)   # Peak consumption

X = np.column_stack((daily, peak))

# Estimate parameters
mean_vector = np.mean(X, axis=0)
cov_matrix = np.cov(X.T)

print(f"Mean Vector: {mean_vector}")
print(f"Covariance Matrix:\n{cov_matrix}")

# Create Gaussian model
rv = multivariate_normal(mean_vector, cov_matrix)

# Create grid for plotting
x, y = np.mgrid[3:7:0.05, 1:3:0.05]
pos = np.dstack((x, y))

# Plot Contour
plt.figure(figsize=(6,5))
plt.contourf(x, y, rv.pdf(pos), levels=20, cmap='viridis')
plt.colorbar(label='Probaility Density')

# Plot data points
plt.scatter(X[:, 0], X[:, 1], c='red', edgecolor='k', label='Data Points')
plt.title("Multivariate Gaussian Distribution")
plt.xlabel("Daily Consumption (kWh)")
plt.ylabel("Peak Consumption (kWh)")
plt.legend()
plt.show()
```
